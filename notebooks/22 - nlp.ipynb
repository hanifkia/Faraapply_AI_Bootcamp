```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('punkt_tab')
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

text = "I was loving the NLP class very much!"

# Tokenization
tokens = word_tokenize(text)

# Stopword Removal
filtered = [word for word in tokens if word.lower() not in stopwords.words('english')]

# Stemming
stemmer = PorterStemmer()
stems = [stemmer.stem(word) for word in filtered]

# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(word) for word in filtered]

print("Tokens:", tokens)
print("Filtered:", filtered)
print("Stems:", stems)
print("Lemmas:", lemmas)
```

## Create Features
### Bag of Words (BoW)

```python
["I love NLP", "NLP is amazing", "I love NLP and NLP is amazing"]
```

```python
from sklearn.feature_extraction.text import CountVectorizer

corpus = ["I love NLP", "NLP is amazing", "I love NLP and NLP is amazing"]
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

### TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```
