```python
import numpy as np
from sklearn.datasets import load_diabetes
```

```python
data = load_diabetes()
xData = data['data']
yData = data['target']
```

```python
from sklearn.model_selection import train_test_split
```

```python
xTrain, xTest, yTrain, yTest = train_test_split(xData, yData, random_state=42)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
x_scaler = MinMaxScaler()
xTrain = x_scaler.fit_transform(xTrain)
xTest = x_scaler.transform(xTest)
```

```python
yTrain.shape
```

```python
y_scaler = MinMaxScaler()
yTrain = y_scaler.fit_transform(np.expand_dims(yTrain, axis=-1))
yTest = y_scaler.transform(np.expand_dims(yTest, axis=-1))
```

```python
yTrain.shape
```

```python
#!pip install tensorflow
```

```python
import tensorflow as tf
from tensorflow.keras import layers
```

```python
input_layer = tf.keras.Input(shape=(xTrain.shape[1:]), name="my_input")
x = layers.Dense(32, activation='sigmoid')(input_layer)
x = layers.Dropout(0.25)(x)
x = layers.Dense(15)(x)
x = layers.Dense(1, activation="linear")(x)

model = tf.keras.Model(input_layer, x)
```

```python
model.summary()
```

```python
model.compile(optimizer=tf.keras.optimizers.SGD(), loss=tf.keras.losses.MeanSquaredError())
```

```python
model.fit(xTrain, yTrain, epochs=20, validation_data=(xTest, yTest))
```

```python
model.predict()
```
