```python
import numpy as np
from sklearn.datasets import load_breast_cancer, load_iris
import tensorflow as tf
from tensorflow.keras import layers
```

```python
from sklearn.model_selection import train_test_split
```

# Binary Classification
### Breast Cancer

```python
data = load_breast_cancer()
xData = data['data']
yData = data["target"]
```

```python
xTrain, xTest, yTrain, yTest = train_test_split(xData, yData, random_state=42)
```

```python
# normalization
...
```

```python
input_layer = tf.keras.Input(shape=xTrain.shape[1:])
model = layers.Dense(64, activation="elu")(input_layer)
model = layers.Dropout(0.2)(model)
model = layers.Dense(32, activation='elu')(model)
model = layers.Dropout(0.2)(model)
model = layers.Dense(1, activation='sigmoid')(model)

model = tf.keras.Model(input_layer, model)
model.summary()
```

```python
model.compile(optimizer=tf.keras.optimizers.Adam(0.01), loss=tf.keras.losses.BinaryCrossentropy())
```

```python
model.fit(xTrain, yTrain, validation_data=(xTest, yTest), epochs=20)
```

```python
from sklearn.metrics import classification_report
```

```python
y_pred = model.predict(xTest)
```

```python
print(classification_report(yTest, (y_pred > 0.5)))
```

# Multiclass
## CCE
### Iris

```python
data = load_iris()
xData = data['data']
yData = data["target"]
```

```python
xTrain, xTest, yTrain, yTest = train_test_split(xData, yData, random_state=42)
```

```python
# normalization
# ....
```

```python
yTrain
```

```python
np.unique(yTrain)
```

```python
input_layer = tf.keras.Input(shape=xTrain.shape[1:])
model = layers.Dense(32, activation="elu")(input_layer)
model = layers.Dropout(0.2)(model)
model = layers.Dense(16, activation='elu')(model)
model = layers.Dropout(0.2)(model)
model = layers.Dense(len(np.unique(yTrain)), activation='softmax')(model)

model = tf.keras.Model(input_layer, model)
model.summary()
```

```python
model.compile(optimizer=tf.keras.optimizers.Adam(0.01), loss=tf.keras.losses.CategoricalCrossentropy())
```

```python
yTrain
```

```python
model.fit(xTrain, tf.keras.utils.to_categorical(yTrain), validation_data=(xTest, tf.keras.utils.to_categorical(yTest)), epochs=20)
```

```python
y_pred = model.predict(xTest)
```

```python
y_pred = np.argmax(y_pred, 1)
```

```python
print(classification_report(yTest, y_pred))
```
