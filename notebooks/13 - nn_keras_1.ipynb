```python
import numpy as np
from sklearn.datasets import load_diabetes
```

```python
data = load_diabetes()
xData = data['data']
yData = data['target']
```

```python
from sklearn.model_selection import train_test_split
```

```python
xTrain, xTest, yTrain, yTest = train_test_split(xData, yData, random_state=42)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
x_scaler = MinMaxScaler()
xTrain = x_scaler.fit_transform(xTrain)
xTest = x_scaler.transform(xTest)
```

```python
yTrain.shape
```

```python
y_scaler = MinMaxScaler()
yTrain = y_scaler.fit_transform(np.expand_dims(yTrain, axis=-1))
yTest = y_scaler.transform(np.expand_dims(yTest, axis=-1))
```

```python
yTrain.shape
```

```python
#!pip install tensorflow
```

```python
import tensorflow as tf
from tensorflow.keras import layers
```

```python
model = tf.keras.Sequential()
model.add(tf.keras.Input(shape=xTrain.shape[1:]))
model.add(layers.Dense(8, activation='sigmoid'))
model.add(layers.Dense(16, activation='sigmoid'))
model.add(layers.Dense(1, activation='linear'))
```

```python
# model.compile(optimizer='sgd', loss="mean_squared_error")
model.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=0.001), loss=tf.keras.losses.MeanSquaredError())
```

```python
model.summary()
```

```python
model.fit(x=xTrain, y=yTrain, epochs=5, validation_data=(xTest, yTest), verbose=1, batch_size=64)
```

```python
y_pred = model.predict(xTest)
```

```python
from sklearn.metrics import mean_squared_error
```

```python
mean_squared_error(yTest, y_pred)
```
