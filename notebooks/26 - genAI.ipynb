## What is Generative AI?

**Definition**: AI systems that create new content from learned patterns

**Key Distinction**:
- **Discriminative Models**: P(y|x) - "What class is this?"
- **Generative Models**: P(x) or P(x|y) - "Create new examples"

**Applications**:
- Image/Video generation (DALL-E, Midjourney)
- Text generation (GPT, ChatGPT)
- Audio synthesis (music, speech)
- Code generation (GitHub Copilot)

---

## The Generative Model Zoo

**Major Approaches**:
1. **Variational Autoencoders (VAEs)** - Probabilistic encoding
2. **Generative Adversarial Networks (GANs)** - Adversarial training
3. **Diffusion Models** - Iterative denoising
4. **Autoregressive Models** - Sequential generation
5. **Flow-based Models** - Invertible transformations
6. **Transformer-based** - Attention mechanisms

**Evolution**: VAEs → GANs → Diffusion → Transformers


## Variational Autoencoders (VAEs)
### The Probabilistic Pioneer
![alt text](vae.webp "Title")

**Core Idea**: Learn continuous latent space with probabilistic encoder/decoder

**Architecture**:
- **Encoder**: x → μ, σ (mean & variance)
- **Latent Space**: z ~ N(μ, σ²)
- **Decoder**: z → x'

**Key Innovation**: Reparameterization trick
```
z = μ + σ ⊙ ε, where ε ~ N(0,1)
```

## VAE: Strengths & Limitations

**✅ Strengths**:
- Stable training
- Meaningful latent interpolation
- Theoretical foundation
- Good for representation learning

**❌ Limitations**:
- Blurry outputs
- Limited generation quality
- Posterior collapse issues

**Use Cases**:
- Data compression
- Anomaly detection
- Latent space exploration
- Semi-supervised learning

## Generative Adversarial Networks (GANs)
### The Game-Theoretic Revolution

**Core Concept**: Two networks in adversarial competition

**Players**:
- **Generator (G)**: Creates fake data to fool discriminator
- **Discriminator (D)**: Distinguishes real from fake data

**Training Process**:
```
min_G max_D V(D,G) = E[log D(x)] + E[log(1-D(G(z)))]
```

**Analogy**: Counterfeiter (G) vs Police (D)

## GAN Training Dynamics

**The Training Dance**:
1. **D Training**: Learn to distinguish real/fake
2. **G Training**: Learn to fool discriminator
3. **Equilibrium**: G creates realistic data, D can't distinguish

**Nash Equilibrium**: Both networks reach optimal strategy

**Training Challenges**:
- Mode collapse
- Training instability
- Vanishing gradients

## GAN Evolution Timeline

**2014**: Original GAN (Goodfellow)
**2015**: DCGAN - Convolutional architecture
**2016**: InfoGAN - Controllable generation
**2017**: WGAN - Wasserstein distance
**2018**: BigGAN - High-resolution images
**2019**: StyleGAN - Style transfer control
**2021**: StyleGAN3 - Alias-free generation

**Key Insight**: Architecture + training tricks = better results
